# Initial analysis of data

```python
# Cell 1: setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/case1Data.csv")

sns.set_theme(style="whitegrid", palette="colorblind")
pd.set_option("display.max_columns", None)
```

```python
# Cell 2: load data
na_tokens = ["NaN", ""]
df = pd.read_csv(DATA_PATH, na_values=na_tokens)
target = "y"

print(f"Rows: {df.shape[0]:,}  Columns: {df.shape[1]:,}")
df.head()
```

```python
# Cell 3: dtype & missingness summary
info = pd.DataFrame({
    "dtype": df.dtypes,
    "n_missing": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(1),
})
info.sort_values("missing_pct", ascending=False)
```

```python
# Cell 4: visualize missingness (top 25)
missing = df.isna().sum().sort_values(ascending=False)
missing = missing[missing > 0]
if missing.empty:
    print("No missing values detected.")
else:
    top = missing.head(25)
    plt.figure(figsize=(8, 10))
    sns.barplot(x=top.values / len(df) * 100, y=top.index, color="#4477aa")
    plt.xlabel("% of rows missing")
    plt.ylabel("Feature")
    plt.title("Top missing features")
    plt.tight_layout()
```

```python
# Cell 5: target distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df[target], kde=True, ax=axes[0])
axes[0].set_title(f"{target} distribution")
sns.boxplot(x=df[target], ax=axes[1])
axes[1].set_title(f"{target} boxplot")
plt.tight_layout()

df[target].describe()
```

```python
# Cell 6: split features by prefix (heuristic)
num_cols = [c for c in df.columns if c.startswith("x_")]
cat_cols = [c for c in df.columns if c.startswith("C_")]

print(f"Numeric features: {len(num_cols)} | Categorical-like: {len(cat_cols)}")
if cat_cols:
    display(df[cat_cols].describe(include="all").T)
```

```python
# Cell 8: correlation of numeric features with target
corr_to_y = df[num_cols].corrwith(df[target])
corr_sorted = corr_to_y.dropna().reindex(corr_to_y.abs().sort_values(ascending=False).index)

top_corr = corr_sorted.head(15)
print(top_corr)

plt.figure(figsize=(8, 6))
sns.barplot(x=top_corr.values, y=top_corr.index, palette="crest")
plt.title("Top correlations with target y")
plt.tight_layout()
```

```python
# Cell 9: scatterplots for the strongest signals
if not top_corr.empty:
    top_feats = top_corr.index[:3]
    fig, axes = plt.subplots(1, len(top_feats), figsize=(15, 4))
    for ax, col in zip(axes, top_feats):
        sns.scatterplot(data=df, x=col, y=target, alpha=0.5, ax=ax)
        ax.set_title(f"{col} vs {target}")
    plt.tight_layout()
```
