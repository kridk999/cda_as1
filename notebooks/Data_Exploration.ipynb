# Initial analysis of data

```python
# Cell 1: setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/case1Data.csv")

sns.set_theme(style="whitegrid", palette="colorblind")
pd.set_option("display.max_columns", None)
```

```python
# Cell 2: load data
na_tokens = ["NaN", ""]
df = pd.read_csv(DATA_PATH, na_values=na_tokens)
target = "y"

print(f"Rows: {df.shape[0]:,}  Columns: {df.shape[1]:,}")
df.head()
```

```python
# Cell 3: dtype & missingness summary
info = pd.DataFrame({
    "dtype": df.dtypes,
    "n_missing": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(1),
})
info.sort_values("missing_pct", ascending=False)
```

```python
# Cell 4: visualize missingness (top 25)
missing = df.isna().sum().sort_values(ascending=False)
missing = missing[missing > 0]
if missing.empty:
    print("No missing values detected.")
else:
    top = missing.head(25)
    plt.figure(figsize=(8, 10))
    sns.barplot(x=top.values / len(df) * 100, y=top.index, color="#4477aa")
    plt.xlabel("% of rows missing")
    plt.ylabel("Feature")
    plt.title("Top missing features")
    plt.tight_layout()
```

```python
# Cell 5: target distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df[target], kde=True, ax=axes[0])
axes[0].set_title(f"{target} distribution")
sns.boxplot(x=df[target], ax=axes[1])
axes[1].set_title(f"{target} boxplot")
plt.tight_layout()

df[target].describe()
```

```python
# Cell 6: split features by prefix (heuristic)
num_cols = [c for c in df.columns if c.startswith("x_")]
cat_cols = [c for c in df.columns if c.startswith("C_")]

print(f"Numeric features: {len(num_cols)} | Categorical-like: {len(cat_cols)}")
if cat_cols:
    display(df[cat_cols].describe(include="all").T)
```

```python
# Cell 8: correlation of numeric features with target
corr_to_y = df[num_cols].corrwith(df[target])
corr_sorted = corr_to_y.dropna().reindex(corr_to_y.abs().sort_values(ascending=False).index)

top_corr = corr_sorted.head(15)
print(top_corr)

plt.figure(figsize=(8, 6))
sns.barplot(x=top_corr.values, y=top_corr.index, palette="crest")
plt.title("Top correlations with target y")
plt.tight_layout()
```

```python
# Cell 9: scatterplots for the strongest signals
if not top_corr.empty:
    top_feats = top_corr.index[:3]
    fig, axes = plt.subplots(1, len(top_feats), figsize=(15, 4))
    for ax, col in zip(axes, top_feats):
        sns.scatterplot(data=df, x=col, y=target, alpha=0.5, ax=ax)
        ax.set_title(f"{col} vs {target}")
    plt.tight_layout()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Lars, LassoLars, ElasticNetCV, LassoLarsCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
import matplotlib.pyplot as plt

# ── 1. Define feature types ───────────────────────────────────────────────────
target_col = 'y'

# Match your actual column naming convention
categorical_cols = [c for c in df.columns if c.startswith('C_')]
continuous_cols  = [c for c in df.columns if c.startswith('x_')]

print(f"Continuous features: {len(continuous_cols)}")
print(f"Categorical features: {len(categorical_cols)}")

X = df.drop(columns=[target_col])
y = df[target_col]

```

```python

# ── 2. Preprocessing ──────────────────────────────────────────────────────────
# LARS/ElasticNet require:
#   - Scaled continuous features (zero mean, unit variance)
#   - Encoded categorical features
#   - No missing values

preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(),                        continuous_cols),
    ('cat', OneHotEncoder(drop='first', sparse_output=False), categorical_cols),
])

X_processed = preprocessor.fit_transform(X)

# Recover feature names after encoding
ohe_feature_names = preprocessor.named_transformers_['cat']\
                                 .get_feature_names_out(categorical_cols).tolist()
all_feature_names = continuous_cols + ohe_feature_names

# Handle missing values (LARS does not handle NaN natively)
X_df = pd.DataFrame(X_processed, columns=all_feature_names)
X_df = X_df.fillna(X_df.median())
X_processed = X_df.values
```

```python
X_processed
```

```python

# ── 3a. LARS — traces the full coefficient path ───────────────────────────────
lars = Lars(n_nonzero_coefs=20)  # limit to 20 nonzero for speed
lars.fit(X_processed, y)

lars_selected = pd.Series(lars.coef_, index=all_feature_names)
lars_selected = lars_selected[lars_selected != 0].abs().sort_values(ascending=False)
print("LARS selected features:")
print(lars_selected)

# ── 3b. LassoLARS with Cross-Validation (finds optimal alpha) ─────────────────
lasso_lars_cv = LassoLarsCV(cv=5, max_iter=500)
lasso_lars_cv.fit(X_processed, y)

lasso_lars_coefs = pd.Series(lasso_lars_cv.coef_, index=all_feature_names)
lasso_lars_selected = lasso_lars_coefs[lasso_lars_coefs != 0]\
                        .abs().sort_values(ascending=False)
print(f"\nLassoLARS CV selected {len(lasso_lars_selected)} features:")
print(lasso_lars_selected)

# ── 3c. Elastic Net with Cross-Validation ─────────────────────────────────────
# l1_ratio: 1.0 = pure Lasso, 0.0 = pure Ridge, 0.5 = balanced
enet_cv = ElasticNetCV(
    l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1.0],  # searches over these
    cv=5,
    max_iter=10000,
    random_state=42
)
enet_cv.fit(X_processed, y)

print(f"\nElastic Net best alpha:    {enet_cv.alpha_:.4f}")
print(f"Elastic Net best l1_ratio: {enet_cv.l1_ratio_:.2f}")

enet_coefs = pd.Series(enet_cv.coef_, index=all_feature_names)
enet_selected = enet_coefs[enet_coefs != 0].abs().sort_values(ascending=False)
print(f"\nElastic Net selected {len(enet_selected)} features:")
print(enet_selected)
```

```python

# ── 4. Select Top K features ──────────────────────────────────────────────────
K = 15  # set your desired K

top_k_features = enet_selected.head(K).index.tolist()
print(f"\nTop {K} features selected by Elastic Net:")
print(top_k_features)

X_reduced = X_df[top_k_features]  # final reduced dataframe
```

```python

from sklearn.linear_model import lasso_path

alphas, coefs, _ = lasso_path(X_processed, y.values, eps=1e-3)

plt.figure(figsize=(12, 6))
for i, name in enumerate(all_feature_names):
    if abs(coefs[i, -1]) > 0.01:
        plt.plot(alphas, coefs[i], label=name)

plt.xscale('log')
plt.gca().invert_xaxis()
plt.xlabel('Alpha (regularisation strength)')
plt.ylabel('Coefficient value')
plt.title('Lasso Regularisation Path')
plt.legend(bbox_to_anchor=(1.05, 1), fontsize=7)
plt.tight_layout()
plt.show()
```

```python

# ── 6. Compare methods ────────────────────────────────────────────────────────
comparison = pd.DataFrame({
    'LARS':       lars_selected,
    'LassoLARS':  lasso_lars_selected,
    'ElasticNet': enet_selected,
}).fillna(0)

print("\nFeature importance comparison across methods:")
print(comparison.sort_values('ElasticNet', ascending=False).head(K))
```
